```python
###Building Markov transition matrix
from collections import defaultdict
import numpy as np
import pandas as pd

def build_transition_matrix(sequence):
    # Initialize count matrix
    states = ['A', 'C', 'G', 'T']
    transition_counts = defaultdict(lambda: defaultdict(int))

    # Count transitions
    for i in range(len(sequence) - 1):
        curr = sequence[i]
        next_ = sequence[i + 1]
        if curr in states and next_ in states:
            transition_counts[curr][next_] += 1

    # Convert to probability matrix
    matrix = pd.DataFrame(index=states, columns=states, dtype=float)

    for curr in states:
        total = sum(transition_counts[curr].values())
        for next_ in states:
            if total > 0:
                matrix.at[curr, next_] = transition_counts[curr][next_] / total
            else:
                matrix.at[curr, next_] = 0.0

    return matrix

# Example usage
sequence = "ACGTTGCAAGTACGTAGC"  # Replace with your input sequence
transition_matrix = build_transition_matrix(sequence)

# Print the matrix
print("Transition Matrix:")
print(transition_matrix.round(3))
```
